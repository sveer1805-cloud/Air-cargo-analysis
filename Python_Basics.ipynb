# Break Statement

```python
# example that uses break statement in a for Loop

# take range (10) and print ' The number is' + value
# break when num equa;s 5
for num in range(10):
    print("The number is",num)
    if num==5:
        break
```

#Conntinue Statement

```python
# Using same for loop prohram as in Break Statement section above
# Use a continue statement rather than a break statement

# take range(10) and print 'The number is ' + value
#continue when num equalss 5

for num in range (10):
  if num==5:
     continue;
     print("the number is", num )
```
